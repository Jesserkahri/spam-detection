# I. Problématique

Avec l’augmentation des communications numériques, les messages indésirables (spam) représentent un problème majeur pour les utilisateurs et les entreprises. Ces messages peuvent contenir des publicités abusives, des tentatives de fraude ou des contenus malveillants. Il devient donc essentiel de mettre en place des systèmes capables de détecter automatiquement ces messages afin de protéger les utilisateurs et améliorer la qualité des services de communication.

Dans ce projet, nous proposons d’utiliser des techniques de Machine Learning pour classifier automatiquement les messages SMS en deux catégories : spam et non spam (ham). L’objectif est également de comparer plusieurs méthodes de classification afin d’identifier la plus performante.

# II. État de l’art


La détection de spam SMS est un problème classique de classification de texte en machine learning. Elle consiste à identifier automatiquement si un message est légitime (ham) ou indésirable (spam). Plusieurs approches ont été proposées dans la littérature, allant des méthodes probabilistes simples aux modèles plus avancés.

## Méthodes existantes

### 1. Naive Bayes
Naive Bayes est une méthode probabiliste basée sur le théorème de Bayes. Elle suppose que les caractéristiques (mots) sont indépendantes les unes des autres. Malgré cette hypothèse simplificatrice, elle est très efficace pour la classification de texte, notamment pour la détection de spam SMS, en raison de sa rapidité et de sa simplicité.

### 2. Support Vector Machine (SVM)
Le Support Vector Machine est une méthode de classification qui cherche à trouver une frontière optimale séparant les différentes classes. Elle est particulièrement adaptée aux données de grande dimension comme les textes, et offre généralement de bonnes performances en termes de précision.

### 3. Dendritic Cell Algorithm (DCA)
Le Dendritic Cell Algorithm est une méthode inspirée du système immunitaire humain. Elle utilise des signaux pour détecter des anomalies dans les données. Cette approche est utilisée dans certains systèmes de détection d’intrusion et peut être appliquée à la détection de spam en identifiant les comportements anormaux.

### 4. K-Nearest Neighbors (KNN)
KNN est une méthode basée sur la similarité entre les données. Un message est classé en fonction des classes de ses voisins les plus proches. Cette méthode est simple mais peut être coûteuse en temps de calcul pour de grandes bases de données.

### 5. Voted Perceptron
Le Voted Perceptron est une extension de l’algorithme perceptron classique. Il améliore la robustesse du modèle en combinant plusieurs classifieurs. Il est utilisé pour la classification de texte grâce à sa capacité à gérer des données linéairement séparables.

### 6. AdaBoost
AdaBoost est une méthode d’ensemble qui combine plusieurs classifieurs faibles pour créer un modèle plus performant. Elle accorde plus d’importance aux exemples mal classés, ce qui permet d’améliorer progressivement la précision du modèle.

### 7. Lazy K-Star (K*)
K-Star est une méthode basée sur l’apprentissage paresseux (lazy learning). Elle utilise une fonction de distance basée sur l’entropie pour comparer les instances. Elle est moins utilisée mais peut donner de bons résultats dans certains cas de classification.

### 8. Hybrid Methods
Les méthodes hybrides combinent plusieurs algorithmes afin de tirer parti de leurs avantages respectifs. Par exemple, combiner Naive Bayes et SVM permet d’améliorer la performance globale du système de détection de spam.

### 9. Evolutionary Classifier
Les classifieurs évolutionnaires sont basés sur des algorithmes génétiques. Ils permettent d’optimiser automatiquement les paramètres du modèle ou de sélectionner les meilleures caractéristiques, ce qui peut améliorer les performances de classification.

### 10. Large Cellular Network
Les approches basées sur les réseaux cellulaires utilisent des structures distribuées pour traiter les données. Elles sont inspirées des systèmes biologiques et permettent de modéliser des interactions complexes entre les données.

### 11. Graph-based KNN
Le Graph-based KNN est une extension de KNN qui utilise des graphes pour représenter les relations entre les données. Cette approche permet d’améliorer la classification en exploitant la structure globale des données plutôt que des comparaisons locales uniquement.

### 12. Random Forest
Random Forest est un ensemble d’arbres de décision. Elle permet de capturer des relations complexes dans les données et améliore la robustesse du modèle.

### 13. Régression Logistique
La régression logistique est un modèle linéaire utilisé pour la classification binaire. Elle permet de prédire la probabilité qu’un message soit un spam.


## Synthèse

Les méthodes classiques comme Naive Bayes et la régression logistique offrent des performances satisfaisantes avec une faible complexité. Cependant, les modèles plus avancés comme SVM et Random Forest permettent généralement d’obtenir de meilleurs résultats. Les approches de deep learning offrent des performances supérieures mais nécessitent plus de données et de ressources de calcul.

# III. Méthodologie

## 1. Démarche générale

Dans ce projet, nous avons adopté une démarche expérimentale visant à concevoir un système de détection de spam SMS basé sur des techniques de machine learning supervisé.

Le processus a commencé par l’utilisation d’un dataset contenant environ 5500 messages SMS, chacun étiqueté comme spam ou ham. Ce problème correspond à une classification binaire, où l’objectif est de prédire si un message appartient à la classe des messages indésirables ou à celle des messages normaux.

Une phase de prétraitement a été appliquée aux messages afin de nettoyer les données textuelles. Cette étape comprend la conversion des textes en minuscules, la suppression des caractères spéciaux ainsi que l’élimination des mots vides. Les textes nettoyés ont ensuite été transformés en vecteurs numériques grâce à la méthode TF-IDF, qui attribue un poids à chaque mot selon son importance dans le corpus.

Les données ont ensuite été divisées en deux ensembles : un ensemble d’entraînement représentant 80% des données et un ensemble de test représentant 20%. Cette séparation permet d’entraîner les modèles puis d’évaluer leur capacité de généralisation sur des données non vues.

Quatre modèles de classification supervisée ont été utilisés dans cette étude : Naive Bayes, Régression Logistique, Support Vector Machine et Random Forest. Les performances de ces modèles ont été comparées à l’aide des métriques accuracy, précision, recall et F1-score.

## 2. Description des modèles utilisés

### a. Naive Bayes

Le modèle Naive Bayes est un classificateur probabiliste basé sur le théorème de Bayes. Il suppose que les caractéristiques sont indépendantes entre elles, ce qui simplifie le calcul des probabilités.

La probabilité d’appartenance d’un message à une classe est donnée par :

$$
P(C \mid X) = \frac{P(X \mid C) \cdot P(C)}{P(X)}
$$

où :
- $P(C \mid X)$ représente la probabilité qu’un message $X$ appartienne à la classe $C$
- $P(X \mid C)$ est la probabilité d’observer $X$ sachant la classe $C$
- $P(C)$ est la probabilité a priori de la classe

Le message est affecté à la classe ayant la probabilité maximale.

---

### b. Régression Logistique

La régression logistique est un modèle de classification qui estime la probabilité qu’une observation appartienne à une classe donnée.

Elle utilise la fonction logistique :

$$
P(Y=1 \mid X) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 X_1 + \dots + \beta_n X_n)}}
$$

où :
- $\beta_0$ est la constante
- $\beta_1$ à $\beta_n$ sont les coefficients des variables explicatives

Si la probabilité calculée est supérieure à un seuil (généralement 0.5), le message est classé comme spam.

---

### c. Support Vector Machine (SVM)

Le modèle SVM cherche à trouver l’hyperplan optimal qui sépare les classes en maximisant la marge entre les observations des deux classes.

L’équation de l’hyperplan est :

$$
w \cdot x + b = 0
$$

où :
- $w$ est le vecteur normal à l’hyperplan
- $x$ est le vecteur de caractéristiques
- $b$ est le biais

L’objectif est de maximiser la distance entre l’hyperplan et les points les plus proches appelés vecteurs de support.

---

### d. Random Forest

Random Forest est une méthode d’ensemble qui combine plusieurs arbres de décision afin d’améliorer la performance globale du modèle.

Chaque arbre produit une prédiction, puis la décision finale est obtenue par vote majoritaire :

$$
\hat{y} = \mathrm{mode}(T_1(x), T_2(x), \dots, T_n(x))
$$

où :
- $T_1$ à $T_n$ représentent les arbres de décision
- $\mathrm{mode}$ désigne la classe la plus fréquemment prédite

Cette méthode permet de réduire le surapprentissage et d’améliorer la robustesse du modèle.

# 4. Implémentation

L’implémentation de ce projet a été réalisée sous l’environnement Google Colab en utilisant le langage Python et la bibliothèque scikit-learn.

Le processus d’implémentation a été organisé en plusieurs étapes :

### 1. Recherche et acquisition du dataset
Le dataset utilisé est le "SMS Spam Collection Dataset", récupéré depuis la plateforme Kaggle. Ce dataset contient environ 5500 messages SMS étiquetés comme spam ou ham.

### 2. Chargement des données
Le fichier CSV a été importé dans l’environnement Google Colab, puis chargé à l’aide de la bibliothèque pandas afin de permettre sa manipulation.

### 3. Préparation des données
Les colonnes utiles ont été sélectionnées et renommées en "label" et "message" afin de simplifier leur utilisation.

### 4. Analyse exploratoire
Une analyse initiale a été effectuée pour vérifier la distribution des classes (spam et ham) et comprendre le déséquilibre du dataset.

### 5. Prétraitement du texte
Les messages ont été nettoyés en appliquant plusieurs transformations :
- conversion en minuscules
- suppression des caractères spéciaux
- suppression des mots vides (stop words)

Cette étape permet d’améliorer la qualité des données textuelles.

### 6. Vectorisation des données
Les messages ont été transformés en vecteurs numériques à l’aide de la méthode TF-IDF (Term Frequency - Inverse Document Frequency), permettant de représenter l’importance des mots dans chaque message.

### 7. Encodage des labels
Les étiquettes textuelles (ham, spam) ont été converties en valeurs numériques (0, 1) afin de pouvoir être utilisées par les modèles de machine learning.

### 8. Division des données
Les données ont été séparées en deux ensembles :
- 80% pour l’entraînement
- 20% pour le test

Cette division a été réalisée à l’aide de la fonction train_test_split.

### 9. Entraînement des modèles
Quatre modèles de classification ont été entraînés :
- Naive Bayes (MultinomialNB)
- Régression Logistique (LogisticRegression)
- Support Vector Machine (SVM)
- Random Forest (RandomForestClassifier)

Chaque modèle a été ajusté sur les données d’entraînement.

### 10. Évaluation des performances
Les modèles ont été évalués sur les données de test à l’aide des métriques suivantes :
- Accuracy
- Précision
- Recall
- F1-score

Ces métriques permettent de mesurer la qualité de classification, en particulier pour la détection des messages spam.

### 11. Comparaison des modèles
Les résultats obtenus ont été comparés afin d’identifier le modèle le plus performant dans le contexte de la détection de spam SMS.

Le code complet de cette implémentation est disponible dans le notebook Google Colab associé au projet.

# V. Résultats

Les différents modèles ont été évalués à l’aide de plusieurs métriques : l’accuracy, la précision, le recall et le F1-score.

Les résultats obtenus montrent que tous les modèles offrent de bonnes performances globales, avec une accuracy supérieure à 95%. Cependant, des différences apparaissent lorsqu’on analyse plus en détail la détection des messages spam.

Le modèle Naive Bayes obtient une accuracy de 96%, avec une excellente détection des messages normaux (recall = 1.00), mais un recall plus faible pour les messages spam (0.72), ce qui signifie qu’une partie des spams n’est pas détectée.

La régression logistique présente des performances similaires, avec une légère amélioration du recall pour les spams (0.75), mais reste limitée dans la détection complète des messages indésirables.

Le modèle Random Forest améliore davantage les résultats, avec un recall de 0.83 pour la classe spam, ce qui indique une meilleure capacité à détecter les messages indésirables.

Enfin, le modèle SVM obtient les meilleures performances globales, avec une accuracy de 98% et un recall de 0.87 pour la classe spam. Il offre ainsi le meilleur compromis entre précision et capacité de détection.

Il est également important de noter que tous les modèles présentent une précision très élevée pour la classe spam (proche de 1.00), ce qui signifie que les faux positifs sont très rares.

# VI. Comparaison

Une analyse comparative des modèles permet de mettre en évidence leurs forces et leurs limites.

Naive Bayes est un modèle simple, rapide et efficace pour une première approche. Cependant, son hypothèse d’indépendance entre les variables limite sa capacité à détecter tous les spams, comme le montre son recall relativement faible.

La régression logistique offre des performances légèrement supérieures à Naive Bayes, tout en restant simple à implémenter. Elle constitue un bon compromis entre simplicité et efficacité.

Random Forest, en tant que modèle d’ensemble, permet de capturer des relations plus complexes dans les données. Cela se traduit par une amélioration du recall pour les messages spam, mais au prix d’une complexité plus élevée.

Le modèle SVM se distingue comme le plus performant. Il offre la meilleure accuracy ainsi que le meilleur recall pour la classe spam. Sa capacité à gérer des données de grande dimension, comme le texte, lui permet de mieux séparer les classes.

Ainsi, dans le cadre de ce projet, SVM apparaît comme le modèle le plus adapté pour la détection de spam SMS.

Cependant, le choix du modèle dépend également des contraintes du système. Si la rapidité est prioritaire, Naive Bayes reste une option intéressante. En revanche, pour maximiser les performances, SVM constitue le meilleur choix.
