# Détection de Spam SMS avec Machine Learning

## Réalisé par : [Kahri Jasser]

# 1. Problématique

Avec l’augmentation des communications numériques, les messages indésirables (spam) représentent un problème majeur pour les utilisateurs et les entreprises. Ces messages peuvent contenir des publicités abusives, des tentatives de fraude ou des contenus malveillants. Il devient donc essentiel de mettre en place des systèmes capables de détecter automatiquement ces messages afin de protéger les utilisateurs et améliorer la qualité des services de communication.

Dans ce projet, nous proposons d’utiliser des techniques de Machine Learning pour classifier automatiquement les messages SMS en deux catégories : spam et non spam (ham). L’objectif est également de comparer plusieurs méthodes de classification afin d’identifier la plus performante.

# 2. État de l’art


La détection de spam SMS est un problème classique de classification de texte en machine learning. Elle consiste à identifier automatiquement si un message est légitime (ham) ou indésirable (spam). Plusieurs approches ont été proposées dans la littérature, allant des méthodes probabilistes simples aux modèles plus avancés.

## Méthodes existantes

### 1. Naive Bayes
Naive Bayes est une méthode probabiliste basée sur le théorème de Bayes. Elle suppose que les caractéristiques (mots) sont indépendantes les unes des autres. Malgré cette hypothèse simplificatrice, elle est très efficace pour la classification de texte, notamment pour la détection de spam SMS, en raison de sa rapidité et de sa simplicité.

### 2. Support Vector Machine (SVM)
Le Support Vector Machine est une méthode de classification qui cherche à trouver une frontière optimale séparant les différentes classes. Elle est particulièrement adaptée aux données de grande dimension comme les textes, et offre généralement de bonnes performances en termes de précision.

### 3. Dendritic Cell Algorithm (DCA)
Le Dendritic Cell Algorithm est une méthode inspirée du système immunitaire humain. Elle utilise des signaux pour détecter des anomalies dans les données. Cette approche est utilisée dans certains systèmes de détection d’intrusion et peut être appliquée à la détection de spam en identifiant les comportements anormaux.

### 4. K-Nearest Neighbors (KNN)
KNN est une méthode basée sur la similarité entre les données. Un message est classé en fonction des classes de ses voisins les plus proches. Cette méthode est simple mais peut être coûteuse en temps de calcul pour de grandes bases de données.

### 5. Voted Perceptron
Le Voted Perceptron est une extension de l’algorithme perceptron classique. Il améliore la robustesse du modèle en combinant plusieurs classifieurs. Il est utilisé pour la classification de texte grâce à sa capacité à gérer des données linéairement séparables.

### 6. AdaBoost
AdaBoost est une méthode d’ensemble qui combine plusieurs classifieurs faibles pour créer un modèle plus performant. Elle accorde plus d’importance aux exemples mal classés, ce qui permet d’améliorer progressivement la précision du modèle.

### 7. Lazy K-Star (K*)
K-Star est une méthode basée sur l’apprentissage paresseux (lazy learning). Elle utilise une fonction de distance basée sur l’entropie pour comparer les instances. Elle est moins utilisée mais peut donner de bons résultats dans certains cas de classification.

### 8. Hybrid Methods
Les méthodes hybrides combinent plusieurs algorithmes afin de tirer parti de leurs avantages respectifs. Par exemple, combiner Naive Bayes et SVM permet d’améliorer la performance globale du système de détection de spam.

### 9. Evolutionary Classifier
Les classifieurs évolutionnaires sont basés sur des algorithmes génétiques. Ils permettent d’optimiser automatiquement les paramètres du modèle ou de sélectionner les meilleures caractéristiques, ce qui peut améliorer les performances de classification.

### 10. Large Cellular Network
Les approches basées sur les réseaux cellulaires utilisent des structures distribuées pour traiter les données. Elles sont inspirées des systèmes biologiques et permettent de modéliser des interactions complexes entre les données.

### 11. Graph-based KNN
Le Graph-based KNN est une extension de KNN qui utilise des graphes pour représenter les relations entre les données. Cette approche permet d’améliorer la classification en exploitant la structure globale des données plutôt que des comparaisons locales uniquement.

### 12. Random Forest
Random Forest est un ensemble d’arbres de décision. Elle permet de capturer des relations complexes dans les données et améliore la robustesse du modèle.

### 13. Régression Logistique
La régression logistique est un modèle linéaire utilisé pour la classification binaire. Elle permet de prédire la probabilité qu’un message soit un spam.


## Synthèse

Les méthodes classiques comme Naive Bayes et la régression logistique offrent des performances satisfaisantes avec une faible complexité. Cependant, les modèles plus avancés comme SVM et Random Forest permettent généralement d’obtenir de meilleurs résultats. Les approches de deep learning offrent des performances supérieures mais nécessitent plus de données et de ressources de calcul.
